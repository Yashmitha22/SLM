```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import stft
```

```python
fs = 1000
t = np.linspace(0, 4, fs*4, endpoint=False)
```

```python
signal = np.concatenate([
    np.sin(2*np.pi*50*t[:1000]),
    np.sin(2*np.pi*150*t[1000:])
])
```

```python
f, time, Zxx = stft(signal, fs, nperseg=256)
```

```python
plt.pcolormesh(time, f, np.abs(Zxx))
plt.title("STFT Spectrogram")
plt.ylabel("Frequency (Hz)")
plt.xlabel("Time (sec)")
plt.show()
```
